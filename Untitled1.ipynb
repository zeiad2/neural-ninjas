```python
!pip install -q tensorflow pillow scikit-learn

import os, json, zipfile, glob, random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from PIL import Image
from sklearn.model_selection import train_test_split
```

```python
ZIP_FILE = "images2.zip"
CSV_FILE = "dataset.csv"
LABEL_MAP_FILE = "label_map.json"

with zipfile.ZipFile(ZIP_FILE, "r") as zip_ref:
    zip_ref.extractall("/content/extracted_dataset")

df = pd.read_csv(CSV_FILE)

with open(LABEL_MAP_FILE, "r") as f:
    label_map = json.load(f)

id_to_class = {v: k for k, v in label_map.items()}
NUM_CLASSES = len(label_map)

print(df.head())
print(label_map)
print("Number of classes:", NUM_CLASSES)
```

```python
PATH_COL = "full_image_path" if "full_image_path" in df.columns else "image_path"

def fix_path(path):
    path = str(path)

    if os.path.exists(path):
        return path

    filename = os.path.basename(path)

    matches = glob.glob(f"/content/extracted_dataset/**/{filename}", recursive=True)

    if len(matches) > 0:
        return matches[0]

    return path

df[PATH_COL] = df[PATH_COL].apply(fix_path)

df = df[df[PATH_COL].apply(os.path.exists)].reset_index(drop=True)

print("Total valid images:", len(df))
print(df.head())
```

```python
if "label" not in df.columns:
    if "class_name" in df.columns:
        df["label"] = df["class_name"].map(label_map)
    else:
        raise ValueError("No label or class_name column found.")

if "class_name" not in df.columns:
    df["class_name"] = df["label"].map(id_to_class)

df["label"] = df["label"].astype(int)

print(df["class_name"].value_counts())
```

```python
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["label"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
IMG_SIZE = 224
BATCH_SIZE = 16

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.08),
    tf.keras.layers.RandomZoom(0.10)
])

def read_image(path):
    img = tf.io.read_file(path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img.set_shape([None, None, 3])
    img = tf.image.resize(img, (IMG_SIZE, IMG_SIZE))
    return img

def load_train_image(path, label):
    img = read_image(path)
    img = data_augmentation(img, training=True)
    img = tf.keras.applications.resnet50.preprocess_input(img)
    return img, label

def load_eval_image(path, label):
    img = read_image(path)
    img = tf.keras.applications.resnet50.preprocess_input(img)
    return img, label

train_ds = tf.data.Dataset.from_tensor_slices(
    (train_df[PATH_COL].values, train_df["label"].values)
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (val_df[PATH_COL].values, val_df["label"].values)
)

test_ds = tf.data.Dataset.from_tensor_slices(
    (test_df[PATH_COL].values, test_df["label"].values)
)

train_ds = train_ds.map(load_train_image).shuffle(500).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.map(load_eval_image).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.map(load_eval_image).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

print("Datasets ready.")
```

```python
base_model = tf.keras.applications.ResNet50(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

base_model.trainable = False

inputs = tf.keras.layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))

x = base_model(inputs, training=False)
x = tf.keras.layers.GlobalAveragePooling2D(name="resnet_feature_vector")(x)
x = tf.keras.layers.Dropout(0.35)(x)
x = tf.keras.layers.Dense(256, activation="relu")(x)
x = tf.keras.layers.Dropout(0.25)(x)

outputs = tf.keras.layers.Dense(NUM_CLASSES, activation="softmax")(x)

cnn_model = tf.keras.Model(inputs, outputs, name="ResNet50_CNN_Classifier")

cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.SparseTopKCategoricalAccuracy(k=3, name="top3_acc")
    ]
)

cnn_model.summary()
```

```python
os.makedirs("/content/models", exist_ok=True)

callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        "/content/models/best_resnet50_cnn.keras",
        monitor="val_accuracy",
        save_best_only=True,
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_accuracy",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]

history = cnn_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=callbacks
)
```

```python
cnn_model = tf.keras.models.load_model("/content/models/best_resnet50_cnn.keras")

test_loss, test_acc, test_top3 = cnn_model.evaluate(test_ds)

print("ResNet50 CNN Test Accuracy:", test_acc)
print("ResNet50 CNN Test Top-3 Accuracy:", test_top3)
```

```python
def cnn_predict(image_path):
    img = read_image(image_path)
    img = tf.keras.applications.resnet50.preprocess_input(img)
    img = tf.expand_dims(img, axis=0)

    pred = cnn_model.predict(img, verbose=0)[0]

    class_id = int(np.argmax(pred))
    class_name = id_to_class[class_id]
    confidence = float(pred[class_id])

    return class_id, class_name, confidence, pred
```

```python
def show_top3_predictions(index):
    row = df.iloc[index]

    image_path = row[PATH_COL]
    true_class = row["class_name"]

    class_id, predicted_class, confidence, pred = cnn_predict(image_path)

    top3_ids = np.argsort(pred)[-3:][::-1]

    img = Image.open(image_path).convert("RGB")

    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.axis("off")
    plt.title(
        f"True: {true_class}\n"
        f"Predicted: {predicted_class} ({confidence:.2%})"
    )
    plt.show()

    print("Index:", index)
    print("Image:", image_path)
    print("True class:", true_class)
    print("Predicted item:", predicted_class)

    print("\nTop 3 CNN predictions:")
    for rank, cid in enumerate(top3_ids, start=1):
        cid = int(cid)
        print(f"{rank}. {id_to_class[cid]}: {pred[cid]:.2%}")
```

```python
feature_extractor = tf.keras.Model(
    inputs=cnn_model.input,
    outputs=cnn_model.get_layer("resnet_feature_vector").output
)

feature_extractor.summary()
```

```python
def extract_resnet_feature(image_path):
    img = read_image(image_path)
    img = tf.keras.applications.resnet50.preprocess_input(img)
    img = tf.expand_dims(img, axis=0)

    feature = feature_extractor.predict(img, verbose=0)
    return feature[0]
```

```python
features = []

for path in df[PATH_COL].values:
    features.append(extract_resnet_feature(path))

features = np.array(features)
all_labels = df["label"].values

print("Features shape:", features.shape)
print("Labels shape:", all_labels.shape)
```

```python
outfit_sequences = [
    ["glasses", "tshirt", "jeans", "sneakers"],
    ["glasses", "hoodie", "jeans", "sneakers"],
    ["glasses", "jacket", "jeans", "sneakers"],

    ["tshirt", "glasses", "jeans", "sneakers"],
    ["tshirt", "jacket", "jeans", "sneakers"],
    ["tshirt", "hoodie", "jeans", "sneakers"],

    ["jeans", "tshirt", "glasses", "sneakers"],
    ["jeans", "hoodie", "glasses", "sneakers"],
    ["jeans", "jacket", "glasses", "sneakers"],

    ["sneakers", "tshirt", "jeans", "glasses"],
    ["sneakers", "hoodie", "jeans", "glasses"],
    ["sneakers", "jacket", "jeans", "glasses"],

    ["hoodie", "jeans", "sneakers", "glasses"],
    ["hoodie", "tshirt", "jeans", "sneakers"],

    ["jacket", "tshirt", "jeans", "sneakers"],
    ["jacket", "glasses", "jeans", "sneakers"],

    ["dress", "glasses", "sneakers"],
    ["dress", "jacket", "glasses", "sneakers"],

    ["shorts", "tshirt", "sneakers", "glasses"],
    ["shorts", "glasses", "tshirt", "sneakers"]
]

# Important:
# We shift input IDs by +1 because 0 is used for padding.
item_to_seq_id = {item: label_map[item] + 1 for item in label_map}
seq_id_to_item = {v: k for k, v in item_to_seq_id.items()}

max_len = 4

X_bilstm = []
y_bilstm = []

for seq in outfit_sequences:
    original_ids = [label_map[item] for item in seq]
    shifted_ids = [item_to_seq_id[item] for item in seq]

    for i in range(1, len(seq)):
        X_bilstm.append(shifted_ids[:i])
        y_bilstm.append(original_ids[i])

X_bilstm = tf.keras.preprocessing.sequence.pad_sequences(
    X_bilstm,
    maxlen=max_len,
    padding="pre"
)

y_bilstm = np.array(y_bilstm)

print("X_bilstm shape:", X_bilstm.shape)
print("y_bilstm shape:", y_bilstm.shape)
print(X_bilstm[:5])
print(y_bilstm[:5])
```

```python
bilstm_model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        input_dim=NUM_CLASSES + 1,
        output_dim=32,
        input_length=max_len,
        mask_zero=True
    ),

    tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(64)
    ),

    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dropout(0.2),

    tf.keras.layers.Dense(NUM_CLASSES, activation="softmax")
])

bilstm_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

bilstm_model.summary()
```

```python
bilstm_callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="loss",
        patience=10,
        restore_best_weights=True,
        verbose=1
    )
]

history_bilstm = bilstm_model.fit(
    X_bilstm,
    y_bilstm,
    epochs=100,
    batch_size=4,
    callbacks=bilstm_callbacks,
    verbose=1
)

bilstm_model.save("/content/models/bilstm_outfit_recommender.keras")

print("BiLSTM recommendation model trained and saved.")
```

```python
def bilstm_recommend(start_item, total_items=4):
    original_sequence = [label_map[start_item]]
    shifted_sequence = [item_to_seq_id[start_item]]

    recommendations = []

    while len(recommendations) < total_items - 1:
        padded = tf.keras.preprocessing.sequence.pad_sequences(
            [shifted_sequence],
            maxlen=max_len,
            padding="pre"
        )

        pred = bilstm_model.predict(padded, verbose=0)[0]

        sorted_ids = np.argsort(pred)[::-1]

        next_original_id = None

        for class_id in sorted_ids:
            class_id = int(class_id)

            if class_id not in original_sequence:
                next_original_id = class_id
                break

        if next_original_id is None:
            break

        original_sequence.append(next_original_id)
        shifted_sequence.append(next_original_id + 1)

        recommendations.append(id_to_class[next_original_id])

    return recommendations
```

```python
print("Input: glasses")
print("BiLSTM recommendation:", bilstm_recommend("glasses"))

print("\nInput: tshirt")
print("BiLSTM recommendation:", bilstm_recommend("tshirt"))

print("\nInput: hoodie")
print("BiLSTM recommendation:", bilstm_recommend("hoodie"))
```

```python
def full_outfit_prediction_resnet_bilstm(index):
    row = df.iloc[index]

    image_path = row[PATH_COL]
    true_class = row["class_name"]

    predicted_class_id, predicted_item, confidence, pred = cnn_predict(image_path)

    resnet_feature = extract_resnet_feature(image_path)

    final_recommendations = bilstm_recommend(predicted_item)

    top3_ids = np.argsort(pred)[-3:][::-1]

    img = Image.open(image_path).convert("RGB")

    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.axis("off")
    plt.title(
        f"True: {true_class}\n"
        f"Predicted: {predicted_item} ({confidence:.2%})"
    )
    plt.show()

    print("Index:", index)
    print("Image:", image_path)
    print("True class:", true_class)
    print("Predicted item:", predicted_item)
    print("Confidence:", f"{confidence:.2%}")
    print("ResNet feature vector shape:", resnet_feature.shape)

    print("\nTop 3 CNN predictions:")
    for rank, cid in enumerate(top3_ids, start=1):
        cid = int(cid)
        print(f"{rank}. {id_to_class[cid]}: {pred[cid]:.2%}")

    print("\nFinal outfit recommendations from BiLSTM:")
    print(final_recommendations)

    return {
        "true_class": true_class,
        "predicted_item": predicted_item,
        "confidence": confidence,
        "feature_shape": resnet_feature.shape,
        "top3": [(id_to_class[int(cid)], float(pred[int(cid)])) for cid in top3_ids],
        "final_recommendations": final_recommendations
    }
```
