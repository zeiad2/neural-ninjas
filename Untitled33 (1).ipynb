```python
import os, json, random
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

# unzip + load data

```python
import zipfile

with zipfile.ZipFile("/content/dataset.zip", "r") as zip_ref:
    zip_ref.extractall("/content")

CSV_PATH = "/content/dataset.csv"
LABEL_MAP_PATH = "/content/label_map.json"
IMAGE_ROOT = "/content/images"

df = pd.read_csv(CSV_PATH)

with open(LABEL_MAP_PATH, "r") as f:
    label_map = json.load(f)

NUM_CLASSES = len(label_map)

print(df.head())
print("Classes:", NUM_CLASSES)
```

# TEXT TOKENIZER + VOCAB

```python
def tokenize(text):
    return str(text).lower().split()

all_words = []
for t in df["text"]:
    all_words.extend(tokenize(t))

vocab = {"<PAD>": 0, "<UNK>": 1}

for w in sorted(set(all_words)):
    vocab[w] = len(vocab)

MAX_LEN = 30

def encode_text(text):
    tokens = tokenize(text)
    ids = [vocab.get(w, vocab["<UNK>"]) for w in tokens]

    if len(ids) < MAX_LEN:
        ids += [0] * (MAX_LEN - len(ids))
    else:
        ids = ids[:MAX_LEN]

    return torch.tensor(ids, dtype=torch.long)

print("Vocab size:", len(vocab))
```

# DATASET

```python
class FashionDataset(Dataset):
    def __init__(self, df, image_root, transform=None):
        self.df = df.reset_index(drop=True)
        self.image_root = image_root
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        img_path = os.path.join("/content", row["image_path"])
        image = Image.open(img_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        text = encode_text(row["text"])
        label = torch.tensor(row["label"], dtype=torch.long)

        return image, text, label
```

# SPLIT + TRANSFORMS

```python
train_df, val_df = train_test_split(
    df,
    test_size=0.3,
    random_state=42,
    stratify=df["label"]
)

train_tfms = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],
                         [0.229,0.224,0.225])
])

val_tfms = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],
                         [0.229,0.224,0.225])
])

train_ds = FashionDataset(train_df, IMAGE_ROOT, train_tfms)
val_ds = FashionDataset(val_df, IMAGE_ROOT, val_tfms)

train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=16)
```

# (ResNet50 + BiGRU)

```python
class MultiModalNet(nn.Module):
    def __init__(self, vocab_size, num_classes):
        super().__init__()


        self.cnn = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        cnn_features = self.cnn.fc.in_features
        self.cnn.fc = nn.Identity()


        self.embedding = nn.Embedding(vocab_size, 64, padding_idx=0)

        self.bigru = nn.GRU(
            input_size=64,
            hidden_size=128,
            batch_first=True,
            bidirectional=True
        )


        self.classifier = nn.Sequential(
            nn.Linear(cnn_features + 256, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

    def forward(self, images, text):

        img_feat = self.cnn(images)

        emb = self.embedding(text)
        _, hidden = self.bigru(emb)

        text_feat = torch.cat([hidden[-2], hidden[-1]], dim=1)

        fused = torch.cat([img_feat, text_feat], dim=1)

        out = self.classifier(fused)

        return out
```

# INIT MODEL

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = MultiModalNet(
    vocab_size=len(vocab),
    num_classes=NUM_CLASSES
).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

print(device)
```

# TRAIN LOOP

```python
EPOCHS = 10

for epoch in range(EPOCHS):
    model.train()

    train_loss = 0
    train_preds, train_labels = [], []

    for images, texts, labels in train_loader:
        images, texts, labels = images.to(device), texts.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images, texts)

        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()

        preds = torch.argmax(outputs, 1)

        train_preds.extend(preds.cpu().numpy())
        train_labels.extend(labels.cpu().numpy())

    train_acc = accuracy_score(train_labels, train_preds)


    model.eval()
    val_preds, val_labels = [], []

    with torch.no_grad():
        for images, texts, labels in val_loader:
            images, texts, labels = images.to(device), texts.to(device), labels.to(device)

            outputs = model(images, texts)
            preds = torch.argmax(outputs, 1)

            val_preds.extend(preds.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    val_acc = accuracy_score(val_labels, val_preds)

    print(f"""
Epoch {epoch+1}/{EPOCHS}
Train Loss: {train_loss/len(train_loader):.4f}
Train Acc : {train_acc:.4f}
Val Acc   : {val_acc:.4f}
""")
```

#REPORT

```python
class_names = list(label_map.keys())

print(classification_report(
    val_labels,
    val_preds,
    target_names=class_names
))
```

# SAVE MODEL

```python
torch.save({
    "model_state_dict": model.state_dict(),
    "vocab": vocab,
    "label_map": label_map
}, "multimodal_resnet_bigru.pth")

print("Saved!")
```
