```python
def find_file(possible_names, pattern):
    for name in possible_names:
        if os.path.exists(name):
            return name

    matches = glob.glob(pattern)
    if len(matches) > 0:
        return matches[0]

    raise FileNotFoundError(f"Could not find file. Tried: {possible_names}")

ZIP_FILE = find_file(
    ["images.zip", "images.zip"],
    "/content/*.zip"
)

CSV_FILE = find_file(
    ["dataset.csv", "dataset.csv"],
    "/content/*.csv"
)

LABEL_MAP_FILE = find_file(
    ["label_map.json", "label_map.json"],
    "/content/*.json"
)

print("ZIP_FILE:", ZIP_FILE)
print("CSV_FILE:", CSV_FILE)
print("LABEL_MAP_FILE:", LABEL_MAP_FILE)

with zipfile.ZipFile(ZIP_FILE, "r") as zip_ref:
    zip_ref.extractall("/content/extracted_dataset")

df = pd.read_csv(CSV_FILE)

with open(LABEL_MAP_FILE, "r") as f:
    label_map = json.load(f)

label_map = {str(k): int(v) for k, v in label_map.items()}
id_to_class = {v: k for k, v in label_map.items()}
NUM_CLASSES = len(label_map)

print(df.head())
print(label_map)
print("Number of classes:", NUM_CLASSES)
```

```python
if "full_image_path" in df.columns:
    PATH_COL = "full_image_path"
elif "image_path" in df.columns:
    PATH_COL = "image_path"
elif "path" in df.columns:
    PATH_COL = "path"
else:
    raise ValueError("No image path column found. Expected full_image_path or image_path.")

def fix_path(path):
    path = str(path)

    if os.path.exists(path):
        return path

    filename = os.path.basename(path)

    matches = glob.glob(f"/content/extracted_dataset/**/{filename}", recursive=True)

    if len(matches) > 0:
        return matches[0]

    return path

df[PATH_COL] = df[PATH_COL].apply(fix_path)
df = df[df[PATH_COL].apply(os.path.exists)].reset_index(drop=True)

print("Total valid images:", len(df))
print(df.head())
```

```python
if "label" not in df.columns:
    if "class_name" in df.columns:
        df["label"] = df["class_name"].map(label_map)
    else:
        raise ValueError("No label or class_name column found.")

if "class_name" not in df.columns:
    df["class_name"] = df["label"].map(id_to_class)

df["label"] = df["label"].astype(int)

print(df["class_name"].value_counts())
```

```python
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["label"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

class_weights_values = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_df["label"]),
    y=train_df["label"]
)

class_weights = {
    int(i): float(w)
    for i, w in zip(np.unique(train_df["label"]), class_weights_values)
}

print(class_weights)
```

```python
IMG_SIZE = 260
BATCH_SIZE = 16

def read_image(path):
    img = tf.io.read_file(path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img.set_shape([None, None, 3])
    img = tf.image.resize(img, (IMG_SIZE, IMG_SIZE))
    return img

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.10),
    tf.keras.layers.RandomZoom(0.15),
    tf.keras.layers.RandomContrast(0.15),
    tf.keras.layers.RandomBrightness(0.10)
])

def load_train_image(path, label):
    img = read_image(path)
    img = data_augmentation(img, training=True)
    img = tf.keras.applications.efficientnet.preprocess_input(img)
    return img, label

def load_eval_image(path, label):
    img = read_image(path)
    img = tf.keras.applications.efficientnet.preprocess_input(img)
    return img, label

train_ds = tf.data.Dataset.from_tensor_slices(
    (train_df[PATH_COL].values, train_df["label"].values)
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (val_df[PATH_COL].values, val_df["label"].values)
)

test_ds = tf.data.Dataset.from_tensor_slices(
    (test_df[PATH_COL].values, test_df["label"].values)
)

train_ds = train_ds.map(load_train_image).shuffle(800).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.map(load_eval_image).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.map(load_eval_image).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

base_model = tf.keras.applications.EfficientNetB2(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

base_model.trainable = False

inputs = tf.keras.layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))

x = base_model(inputs, training=False)
x = tf.keras.layers.GlobalAveragePooling2D(name="efficientnet_feature_vector")(x)
x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.Dropout(0.45)(x)
x = tf.keras.layers.Dense(512, activation="relu")(x)
x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.Dropout(0.35)(x)
x = tf.keras.layers.Dense(256, activation="relu")(x)
x = tf.keras.layers.Dropout(0.25)(x)

outputs = tf.keras.layers.Dense(NUM_CLASSES, activation="softmax")(x)

cnn_model = tf.keras.Model(inputs, outputs, name="EfficientNetB2_CNN")

cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0003),
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.SparseTopKCategoricalAccuracy(k=3, name="top3_acc")
    ]
)

cnn_model.summary()
```

```python
os.makedirs("/content/models", exist_ok=True)

callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        "/content/models/best_efficientnetB2_cnn.keras",
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=8,
        restore_best_weights=True,
        mode="min",
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-6,
        verbose=1
    )
]

history = cnn_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=35,
    callbacks=callbacks,
    class_weight=class_weights
)
```

```python
def cnn_predict(image_path):
    img = read_image(image_path)
    img = tf.keras.applications.efficientnet.preprocess_input(img)
    img = tf.expand_dims(img, axis=0)

    pred = cnn_model.predict(img, verbose=0)[0]

    class_id = int(np.argmax(pred))
    class_name = id_to_class[class_id]
    confidence = float(pred[class_id])

    return class_id, class_name, confidence, pred
```

```python

from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

def get_sample_image_for_class(class_name):
    class_name = str(class_name).lower()

    subset = df[df["class_name"].astype(str).str.lower() == class_name]

    if len(subset) == 0:
        if class_name in label_map:
            subset = df[df["label"].astype(int) == label_map[class_name]]

    if len(subset) == 0:
        return None

    return subset.sample(1).iloc[0][PATH_COL]


def upload_predict_and_show_outfit_images():
    uploaded = files.upload()

    if len(uploaded) == 0:
        print("No image uploaded.")
        return None

    uploaded_image_path = list(uploaded.keys())[0]

    predicted_class_id, predicted_item, confidence, pred = cnn_predict(uploaded_image_path)

    recommended_items = gru_recommend(predicted_item)
    full_outfit = [predicted_item] + recommended_items

    top3_ids = np.argsort(pred)[-3:][::-1]

    print("Uploaded image:", uploaded_image_path)
    print("Predicted item:", predicted_item)
    print("Confidence:", f"{confidence:.2%}")

    print("\nTop 3 CNN predictions:")
    for rank, cid in enumerate(top3_ids, start=1):
        cid = int(cid)
        print(f"{rank}. {id_to_class[cid]}: {pred[cid]:.2%}")

    print("\nFull outfit:")
    for item in full_outfit:
        print("-", item)

    total_images = len(full_outfit)

    plt.figure(figsize=(4 * total_images, 5))

    uploaded_img = Image.open(uploaded_image_path).convert("RGB")

    plt.subplot(1, total_images, 1)
    plt.imshow(uploaded_img)
    plt.axis("off")
    plt.title(f"Uploaded\n{predicted_item}")

    for i, item in enumerate(recommended_items, start=2):
        sample_path = get_sample_image_for_class(item)

        plt.subplot(1, total_images, i)

        if sample_path is not None:
            img = Image.open(sample_path).convert("RGB")
            plt.imshow(img)
            plt.title(item)
        else:
            plt.text(
                0.5,
                0.5,
                f"No image\nfor {item}",
                ha="center",
                va="center"
            )
            plt.title(item)

        plt.axis("off")

    plt.tight_layout()
    plt.show()

    return {
        "uploaded_image": uploaded_image_path,
        "predicted_item": predicted_item,
        "confidence": confidence,
        "top3_predictions": [
            (id_to_class[int(cid)], float(pred[int(cid)]))
            for cid in top3_ids
        ],
        "recommended_items": recommended_items,
        "full_outfit": full_outfit
    }


result = upload_predict_and_show_outfit_images()
```
